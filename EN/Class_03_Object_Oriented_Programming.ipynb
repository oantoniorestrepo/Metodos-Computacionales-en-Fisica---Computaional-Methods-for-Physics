# Introduction to classes and object-oriented programming (OOP)
Python is an interpreted, object-oriented programming language, meaning that
every variable (or list, array, tuple, dictionary) has attributes defined by functions known as methods; for
example `a.sort()`, where `a` is a numpy object and `sort` is an attribute of `a`; `sort` is a method or function.

As mentioned before, every time a new constant or variable is created (this can be a tuple, list, array or dictionary) an object is created which has properties that can be accessed by putting a dot after the variable and the property name, e.g. `x.property()`. In interactive mode, in an ipython cell, you can see the list of attributes by typing the variable followed by a dot and pressing tab; this displays all the methods defined in the class (or library) that can be used — try it with an array and the numpy library.


## Terminology in object-oriented programming (OOP)
Classes are a way of packaging data and functionality together; creating a class creates a new object type, allowing new instances of that type to be created.
Classes are defined in python with the reserved word *class*. Terminology:

**Class variable**: a variable shared by all instances of a class. Class variables are defined within a class but outside any of the class's methods. Class variables are not used as frequently as instance variables.

**Data member**: a class variable or instance variable that holds data associated with a class and its objects.

**Function overloading**: assigning more than one behaviour to a particular function. The operation performed varies according to the types of objects or arguments involved.

**Instance variable**: a variable defined inside a method that belongs only to the current instance of a class.

**Inheritance**: the transfer of the characteristics of a class to other classes derived from it.

**Instance**: an individual object of a given class. An object `obj` that belongs to a class `Circle`, for example, is an instance of the class `Circle`.

**Instantiation**: the creation of an instance of a class.

**Method**: a special kind of function defined within a class definition.

**Object**: a unique instance of a data structure defined by its class. An object comprises both data members (class variables and instance variables) and methods.

**Operator overloading**: assigning more than one function to a particular operator.

(Taken from this [tutorial](https://www.tutorialspoint.com/python3/python_classes_objects.htm)).

Some extra definitions:
* A class is a set of functions and variables with certain encapsulated properties.
* The first argument of a function inside a class is `self`; omitting it will cause an error.
* Method and attribute are more or less the same thing: functions implemented within a class.
* If an object is instantiated from a class, it has the properties of the class (attributes, vars, etc.).
* Object and instance are more or less the same: they are the variables defined from a class.
* Inheritance means the child class inherits the properties of the parent class.
* "Magic methods" are those that start or end with underscores, such as `__init__` (though there are others, like `len`), and they provide special syntactic features or perform special tasks.

Example: below we create the class `atom()` and the class `atomV()`, which inherits the
properties of the class `atom()`:


````python
#%pylab
# parent or base class
class atom:

  # Constructor
  def  __init__(self,x,y,z): # variable initialisation,
      self.x=x               # self is always the first argument
      self.y=y
      self.z=z
  # magic method
  def __abs__(self):         # computing the norm by overloading the abs function
      return (self.x**2. + self.y**2 + self.z**2)**0.5

  # Instance methods
  def coord(self):           # function without arguments (self is mandatory)
      return self.x, self.y, self.z #function that returns the coordinates,

  def E(self,K=1):           # function with one argument.
      return K/(self.x**2. + self.y**2. + self.z**2.)**0.5

a=atom(1,1,1) # a is an object instantiated from the atom class
b=atom(2,2,2) # another object with the same properties as a

print('Coord atoms a,b:',a.coord(), b.coord()) # coord is an attribute or method
print("magnitude a and b", abs(a), abs(b))
````

````python
# child or derived class
class atomV(atom):
    # init all vars
    def  __init__(self,x,y,z, vx,vy,vz,t): # init child class
        atom.__init__(self,x,y,z)          # init parent class
        #super().__init__(x,y,z)           # equivalent to atom.__init__ but more general.

        # init vars of the child class
        self.t=t
        self.vx=vx
        self.vy=vy
        self.vz=vz

    #new functions of the derived class
    def Vel(self):
        return self.x/self.t,self.y/self.t,self.z/self.t

    def Kinetic(self,mass=1):
        Ek=0.5*mass*(self.vx**2.+self.vy**2.+self.vz**2.)
        return Ek


c = atomV(1,1,1,5,5,5,10)
# c is an object with additional attributes compared to a,b
print ('coord Atom',    c.coord()) # Call an instance method and print it
print ('Velocities',    c.Vel())
print ('Potential Ener',c.E())
print ('Kinetic Ener',  c.Kinetic())
````

## What the word "*self*" means and how it is used in Python classes
What is the purpose of the word "*self*" in a Python class?
When an object instance is created, the object itself is passed to the "*self*" parameter,
i.e. "*self*" represents the instance (object) of the class, and using the keyword "*self*" gives access to the class's attributes and methods in Python. For example, suppose a class called `Class` contains a method called `Method`, defined as:
```python
    def Method(self, x, y):
         # do something
```
and `Object = Class()` is an instance of that class; now, when `Object.Method(x, y)` is called, Python internally converts it to:
```python
    Class.Method(Object, x, y)
```
so the variable `self` refers to the object itself. For example, in the earlier `atom` class, the initialisation of the declared object vectors `a, b` takes the form,
```python
     Vector: a = atom(1,1,1)   |  Vector: b = atom(2,2,2)
    ___________________________|__________________________
    def  __init__(a,x,y,z):    |   def  __init__(b,x,y,z):
        a.x = x                |       b.x = x
        a.y = y                |       b.y = y
        a.z = z                |       b.z = z
        a.E(k) = atom.E(a, k)  |       b.E(k) = atom.E(b, k)
```
Note that `a.E(k)` becomes `atom.E(a, k)` and `b.E(k)` becomes `atom.E(b, k)`.

Therefore, "*self*" refers to the bound variable or object; "*self*" is not a reserved keyword in Python, it is simply a strong convention that should always be followed, since avoiding it leads to code errors.

For more, see this [link](https://stackoverflow.com/questions/2709821/what-is-the-purpose-of-self).<br>
For more on "*Magic methods*" see this [link](https://www.tutorialsteacher.com/python/magic-methods-in-python)

## Operator overloading

Consider the operators +,* and the following operations
```python
1 + 1 gives 2
2 * 3 gives 6
'note' + 'book' gives 'notebook'
'Note' * 3 gives 'NoteNoteNote'
```
How can the same operator behave differently depending on context? This is done via operator overloading.
Example:


````python
# https://thepythonguru.com/python-operator-overloading/
# https://www.geeksforgeeks.org/operator-overloading-in-python/

from numpy import pi

class Circ:
    def __init__(self, radius):
        self.__radius = radius # note __radius is different from radius, and that
                             # you could define self.r instead of self.__radius.
    # user-defined functions:
    def Radius(self):
        return self.__radius

    def area(self):
        return pi * self.__radius ** 2

    # these special methods implement operator overloading:
    def __add__(self, other):
        return Circ( self.__radius + other.__radius )

    def __gt__(self, other):
        return self.__radius > other.__radius

    def __lt__(self, other):
        return self.__radius < other.__radius

    def __str__(self):
        return "Radius = " + str(self.__radius)

c1 = Circ(4)
print(c1.Radius())

c2 = Circ(5)
print(c2.Radius())

c3 = c1 + c2    # possible thanks to the __add__ method

print(c3.Radius())
print(c3.area())
print(c3 > c2)  # possible thanks to the __gt__ method
print(c1 < c2)  # possible thanks to the __lt__ method
print(c3)       # possible thanks to the __str__ method
c3
````

## Nested classes
In Python the notation `lib1.lib2.method()` is often used to access methods; this is known as nested classes (or inner classes), i.e. a class is defined inside another class.

To understand how they work, consider the following example, in which we have two nested classes saved to the file `lib1.py` using ipython's magic command `%%writefile lib1.py` (see the appendix):

+ Library1: the main class in the file lib1.py.
+ Library2: a class nested inside Library1.

(Recall that the constructor `__init__` is called every time an instance of the class is created, and `self` is used to refer to the current instance of the class):

Consider the file `lib1.py`, which contains the code:


````python
%%writefile  lib1.py
# File lib1.py
class Library1:
    def __init__(self):
        pass

    def method1(self):
        print("Running method1 in Library1")

    class Library2:
        def __init__(self):
            pass

        def method2(self):
            print("Running method2 in Library2")
````

````python
from lib1 import Library1

library1 = Library1()
library1.method1()  # Prints "Running method1 in Library1"

# Accessing the method inside the nested class
method2 = library1.Library2().method2
method2()  # Prints "Running method2 in Library2"
````

````python
# In fact the name of the object variable doesn't matter:
x = Library1()
x.method1()  # Prints "Running method1 in Library1"

# Accessing the method inside the nested class
y = x.Library2().method2 # Store the function in variable y.
y()                       # Prints "Running method2 in Library2"
x.Library2().method2()   # Note that this does the same as the previous line.
````

## Input for generic functions: args and kwargs
The most general way to pass arguments to a generic function is via args and kwargs; this allows arguments to be passed without needing to know in advance how many values, or which values, will be used. The most general way to pass arguments to a given function is via `*args` and `**kwargs`, which mean passing a tuple/list and a dictionary respectively (note that `args` and `kwargs` are not reserved words, but a widely used convention for defining arguments of generic functions). For example, consider the function,
```python
def f(*a, **b):
    print(a,b)

# let's call the function:
f()          # shows (){}
# but
f(1,2, c=12) # shows (1,2),{'c'=12}
```
that is, `a=(1,2)` is a tuple and `b={'c'=12}` is a dictionary.
This is also used in classes, for example,
```python
class MyClass(FooParentClass):

    def __init__(self, *args, **kwargs):
        print('my function')
        super(MyClass, self).__init__(*args, **kwargs) # init parent class.
```
For example, this method can be used in matplotlib in functions like `plot(*args, **kwargs)`.

Examples:


````python
def f(*a, **b):
    print(a,b)

f(),  f(1,2,c=12,var='hola')
````

## Passing arguments to the main() function
To pass arguments to the `main()` function from the keyboard the `sys` library is required,
```python
import sys

def main():

```
and when the program is run in the terminal,
```bash
$] python my_program.py 1,2 c=12, atom=5
```

this will take the data as input arguments, which are then used within the program; the arguments are stored in the list `sys.argv`, whose first element is the file name. It's important to stress that arguments read via `sys.argv` are stored as a list of strings, so if the arguments given are meant to be numbers they must first be converted to floats or integers; for example you can use `eval(arg)` or `int(arg)` or `float(arg)`, or if the argument is a string like `c=12` you can use `split('=')`, etc. In the following example a script is created, saved to disk, and then run passing it arguments:


````python
%%file Args_main.py

import sys

def main():
    print("Arguments read: ", sys.argv)# show the list of arguments
    for arg in sys.argv[1:]:
        print(arg) # show the arguments one by one (excluding the file name)
        # or do something else with the arguments

if __name__ == "__main__":
    main()
````

## Function decorators
Decorators are a python tool that reduce duplicated lines of code and make code more readable, testable and maintainable. Decorators wrap a function inside another function that returns a function; they are implemented as `@decorator_function_name` before the function to be decorated. In general the syntax is,
```python
@decorate
def foo(a):
    print(a)
```
which is equivalent to writing the code without a decorator,
```python
def bar(a):
    print(a)

foo = decorate(bar)
```
where `decorate(func)` is a previously defined function that receives the function to be decorated as an argument; that is, `decorate(func)` takes a function `func()`, adds some functionality to it, and returns it. Note also that a function can be assigned to a variable, and this variable then becomes another function (i.e. functions are objects in python).
Decorators sometimes take parameters that are passed to the dynamically generated functions to alter their output.
  See more at:
[link 1](https://codigofacilito.com/articulos/decoradores-python)
[link 2](https://www.geeksforgeeks.org/decorators-in-python/),
[link 3](https://www.programiz.com/python-programming/decorator),
[link 4](https://stackoverflow.com/questions/12046883/python-decorator-can-someone-please-explain-this).
So, for example, if a decorator is defined by the function,
```python
def decorate(f):
    f()
    print('Decorated function')
```
then you can do:


````python
# Without a decorator
def decorate(f):
    f()
    print('Decorated function') # functionality added by the decorator

def MyFunction():
    print("Hello")

decorate(MyFunction)  # function without decoration
````

````python
# With a decorator
@decorate
def MyFunction():      # decorated function (can be any function)
    print("Hello")
````

The following example creates a decorator that measures how long a function takes to run:


````python
import math
import time

# ----- Create the decorator function
def measure_time(func):
    '''Measure a function's execution time'''
    def timed(*args, **kwargs):

        begin = time.time()
        func(*args, **kwargs) # Use the generic function (see earlier explanation of args, kwargs).
        end = time.time()
        print("Total time for function", func.__name__, end - begin)

    return timed


#---- Use the decorator --------
@measure_time
def factorial(num):
    time.sleep(2)
    print(math.factorial(num))  # np.math.factorial was removed from NumPy


# Call the decorated function:
factorial(10)
````

### Decorators with input parameters
Decorators can take input parameters; in the following example the function `add()` sums its input arguments, and the decorator `@scale` multiplies the result by a given value. Note that to do this you must define two nested functions: the outer one, which receives the input parameters, `scale(a)`, and the inner one, `f(func)`, which receives the function to be decorated, `func`, as an argument:


````python
 def scale(a=1):
    if (a == 0):
        raise ValueError("a must not be 0")

    def f(func):
        def g(*args, **kwargs):
            return func(*args, **kwargs)*a
        return g
    return f

@scale(a=3)
def add(*args):
    return sum(args)

print(add(1, 3, 5))
print(add(1, 2, 3, 4, 5, 6))
````

## The Walrus operator
Since `python 3.8` another way of assigning variables inside an expression was introduced, `:=`; this operator is known as the [Walrus operator](https://docs.python.org/3/whatsnew/3.8.html#assignment-expressions), and can be used inside `if`, `elif`, `while`, `lists`, `arrays`, etc. Let's see how it is used:


````python
import sys
# First check that the python version is 3.8 or higher
if sys.version_info.major == 3 and sys.version_info.minor >= 8:

    # Ex 1, allows defining n inside the conditional
    if (n:=2>1):print("Ex 1, hello")

    x=[1,2,3,4,5]
    if (n:=len(x)>1):print("Ex 1, hello again")

    # Ex 2, parentheses matter, let's compare:
    if (total := 2 + 5)>1: print("Ex 2,", total) # gives 7,
    if  total := 2 + 5 >1: print("Ex 2,", total) # gives True, because ">1" is included and total becomes logical.

    # Ex 3, run a loop while a random number n generated in [0,10] is less than 8:
    print ("Ex 3, walrus with while:")
    while (n:=np.random.randint(0,10))<8: print(n)

    # Ex 4, reuse an expensive-to-compute value "y" from the function "f(x)"
    x = np.arange(5)
    f = lambda x: 2*x
    print("Ex 4\n", [y := f(x), y**2, y**3]) # builds a 3-row matrix: y, y^2, y^3

    # Ex 5, share a subexpression between a comprehension's filter clause and its output
    data = [False, False, True,1]
    D = [y for x in data if (y := f(x)) is not None]
    print("Ex 5\n",D) # gives [0, 0, 2, 2]

else:
    "Cannot use Walrus, install python 3.8 or higher."
````

## String formatting
There are several ways to insert numeric (or string) data into a string:
```python
string%(value1,value2)            # Format 1
f'string {value1} {value2}... '   # Format 2
string.format(value1, value2...)  # Format 3
```
the values (`value1`,`value2`, etc.) can be given a format. For *format 1* there are several ways to define the format: `%g` is the default format, useful when we don't know the variable's type (logical, string, integer or float), `%d` for integer values, `%f` for floats and `%s` for strings, for example
```python
"integer value %4d"%(n)      # Integer-type format with 4 spaces.
"float value %10.3f"%(x)     # Float-type format with 10 spaces and 3 digits after the point.
"name %d"%(s)                # String-type format.
"unknown value  %g"%(x)      # General format (x can be: logical, string, integer or float).
```
For *format 2* you need to put "f" before the string and the value's format is the default one; for *format 3* the default is `{}` or also `{val}`,
[there are many ways](https://www.w3schools.com/python/ref_string_format.asp) to define `val`; some common ones are,
```python
:c   # Converts the value into the corresponding Unicode character.
:d   # Decimal format.
:e   # Scientific format, with lowercase e.
:E   # Scientific format, with uppercase E.
:f   # Fixed-point number format.
:F   # Fixed-point number format, uppercase (shows inf and nan as INF and NAN).
:g   # General format.
:G   # General format (using an uppercase E for scientific notation).
:o   # Octal format.
:x   # Hexadecimal format, lowercase.
:X   # Hexadecimal format, uppercase.
:n   # Number format.
:%   # Percentage format.
:<   # Left align.
:>   # Right align.
:^   # Centre align.
:=   # Places the sign in the leftmost position.
:+   # Use a plus sign to indicate whether the result is positive or negative.
:-   # Use a minus sign only for negative values.
:    # Use a space to insert an extra space before positive numbers (and a minus sign before negative numbers).
:,   # Use a comma as the thousands separator.
:_   # Use an underscore as the thousands separator.
:b   # Binary format.
```
Let's see some examples:


````python
x=np.pi # Value to insert into the string
print( 'pi value=%g'%(x))    # Way 1, free format.
print( 'pi value=%5.3f'%(x)) # Way 2, float-type format.
print(f'pi value={x}')       # Way 3, x's current format.
print( 'pi value={}'.format(x))
print( 'pi value={:.2f}'.format(x)) # only prints 2 digits after the point.
````

````python
# Three equivalent ways of using format:
s1 = "Name: {fname}, age {age}".format(fname = "John", age = 20)
s2 = "Name: {0}, age {1}".format("Maria",21)
s3 = "Name: {}, age {}".format("George",22)

s1,s2,s3
````

## The subplots, subplot and axes methods
These methods let us make several different plots on the same figure, each with its own axes; below is an explanation and a worked example for each.

The `matplotlib.pyplot.subplots()` method is useful for creating a figure with a set of plots; the syntax is,
```python
fig, ax = subplots(nrows=1, ncols=1, sharex=False, sharey=False, squeeze=True,
             subplot_kw=None, gridspec_kw=None, **fig_kw)
```
where `fig` refers to the figure and is useful, for example, for saving the figure,
```python
 fig.savefig('Figure.png')
```
and `ax` refers to the subplots within the figure; it is a two-dimensional array `(nrows, ncols)` and each element refers to one plot (there are `nrows*ncols` plots), so it can be used, for example, as,
```python
f, ax = plt.subplots(1, 2)         # Method 1
ax[0].plot(x, y1)
ax[0].set_title('My figure 1')
ax[1].plot(x, y2)
```
If there are several plots arranged in a matrix, use `ax[nrows, ncols]`. You can also do:
```python
f, (ax1, ax2) = plt.subplots(1, 2) # Method 2
ax1.plot(x, y1)
ax1.set_title('My figure 1')
```

## Vectorisation
In programming, an operation applied to all elements of an array at once (i.e. all elements are computed simultaneously) is called a [vectorised operation](https://en.wikipedia.org/wiki/Array_programming), regardless of whether it runs on a vector processor (which implements vector instructions, in contrast to serial processes where step two must be completed before step one) or not. Vectorisation is more efficient, since the processor can perform several tasks at once instead of serially, and it is implemented in Python. The Numpy library works with vectorisation, so for example,
```python
c = a + b
```
is more efficient than the serial operation,
```python
for i in range(N):
    c[i] = a[i] + b[i]
```
In python the term "[broadcasting](https://www.tutorialspoint.com/numpy/numpy_broadcasting.htm)" refers to Numpy's ability to handle arrays of different shapes during arithmetic operations; for example if `a` and `b` are Numpy arrays, `c=a+b` is a "broadcasting" operation. Broadcasting also refers to how Numpy treats arrays with different dimensions during arithmetic operations, which leads to certain restrictions; the smaller array is "broadcast" across the larger array so that they have compatible shapes. Broadcasting provides a way of vectorising array operations.


````python
%timeit [math.exp(item) for item in range(150)] # not vectorised
%timeit np.exp(np.arange(150))                  # vectorised
````

### Lists and arrays of functions
In python there are several ways to evaluate a function when the input arguments are lists, tuples or numeric arrays. As we saw earlier, python's "broadcasting" allows a function to be evaluated when the variable is an array, but if the function is user-defined and contains conditionals, broadcasting will fail; for example, a piecewise-defined function such as the absolute value,

$$
f(x)=|x|={\begin{cases}-x,&{\text{if }}x<0\\+x,&{\text{if }}x\geq 0.\end{cases}}
$$

gives the error,
```python
f = lambda x: -x if x<0 else x
a = [1.,2.,3.]
f(a)
TypeError: '<' not supported between instances of 'list' and 'int'
```
How can this function be evaluated on an array? This is where the functions `map()`, `numpy.where()` and `numpy.vectorize()` come in handy.

### Using map()
The python function `map(f, list)` is an iterable that runs a specific function on each element of the iterable. This operation is equivalent to,
```python
[f(x) for x in List ]
```
this is a generic python function and also works with tuples and dictionaries; let's see:


````python
def f(x):
    return 2*x

a = [1,2,3]
# doesn't print, prints a list,  prints a list
map(f,a),     list(map(f,a)), [f(x) for x in a ]
````

````python
a = [-2,0,2]
list(map(lambda x: -x if x<0 else x,a)) # absolute value
````

### Evaluating arrays with numpy.where() and numpy.vectorize()
If the argument is a Numpy array, it is more efficient to use the `numpy.where()` and `numpy.vectorize()` methods, since they apply vectorisation.

The function `np.where(condition,a, b)` returns a Numpy array
whose values will be `a` where the condition is true and `b` where it is false; this is very useful for functions (or arrays) defined with a conditional, such as piecewise functions. If no `a,b` arguments are given, it returns the indices where the condition holds. If `a` and `b` are one-dimensional arrays, `np.where(condition,a, b)` is equivalent to,
 ```python
     [ai if c else bi for c, ai, bi in zip(condition, a, b)]
```
The `numpy.vectorize(func)` method is more general and returns a vectorised function: according to the manual, this method defines a vectorised function that takes a nested sequence of objects or arrays as input and returns an array or a tuple of arrays. The vectorised function evaluates `func` over successive tuples of the input arrays, similar to `map()`, except that it uses Numpy's broadcasting rules and is therefore more efficient. It is useful for generic (user-defined) functions or for functions with more than two pieces; for example consider the function,

$$
f(x)=
\begin{cases}
      0 & x < 0 \\
      1-x & 0\leq x\leq 5 \\
      0 & 5 < x
\end{cases}
$$
implemented as,
```python
def f(x):
    if x<0:           return 0
    elif 0 <= x <= 5: return 1-x
    else:             return 0

x = np.arange(-3,7)

g = np.vectorize(f) # create the vectorisation (returns a function)
y = g(x)            # evaluate the function created by vectorize.
```
Note that the vectorised function must take the same number of arguments as the original function, but as arrays; for example, consider a function of three variables $f(x,a,b)$; then,
```python
a,b = 1,2
x = np.arange(-3,7)

g = np.vectorize(f) # create the vectorisation (returns a function)
y = g(x,a,b)         # evaluate the function created by vectorize.
```
Numerically, `numpy.vectorize(func)` is equivalent to,
```python
y = np.zeros(len(x))      # x is the array of input arguments.
for i,xi in enumerate(x):
    y[i] = f(xi)
```
Examples:


````python
x=np.arange(10)
np.where(x<5,x,0)
````

````python
x = np.array([-2,0,2])
np.where (x<0, -x, x) # absolute value again
````

````python
y = np.random.randint(-10,10,size=10) # Generate 10 random numbers in [-10,10]
np.where(y>0)                         # Indices of y where y>0 holds
````

````python
y[np.where(y>0)]  # Values of y satisfying the condition y>0
````

### Five ways to implement the same piecewise function: efficiency comparison


````python
#%pylab
def f(T): # Function to compute
    if T < 823:
         return -1.206+1.141e-2*T-2.495e-5*T**2+2.461e-8*T**3.-8.873e-12*T**4.
    else:
         return -6.567+3.254e-2*T-5.058e-5*T**2+3.330e-8*T**3.-7.914e-12*T**4.

def P0(T): # 1) With an implicit for.
    y = np.array([f(Ti) for Ti in T])
    return y

def P1(T): # 2) With map()
    y = np.array(list(map(f,T)))
    return y

def P2(T): # 3) With an explicit python for
    y = np.zeros(len(T))
    for i,Ti in enumerate(T):
        y[i] = f(Ti)
    return y

def P3(T): # 4) With numpy.vectorize()
    y = np.vectorize(f)
    return y(T) # Note that y is a function.

def P4(T): # 5) With numpy.where()
    y = np.where(T < 823,
        -1.206+1.141e-2*T-2.495e-5*T**2+2.461e-8*T**3.-8.873e-12*T**4.,
        -6.567+3.254e-2*T-5.058e-5*T**2+3.330e-8*T**3.-7.914e-12*T**4.)
    return y

T = np.arange(100000)

%timeit P0(T) # Very slow.
%timeit P1(T) # Just as slow as P0.
%timeit P2(T) # The slowest.
%timeit P3(T) # One order of magnitude faster, i.e. ~10 times faster.
%timeit P4(T) # The fastest, two orders of magnitude, i.e. ~100 times faster.
````

## More about classes
### Implementing function and method notation using classes
In python libraries you often see a method called either as `lib.method(object)` or, alternatively, as `object.method()`. To implement both ways of calling a method within a class, two techniques are used:

1. Function notation: `lib.method(object)`: used to call a function by passing an object (e.g. an array) as an argument.
2. Method notation: `object.method()`: used to call a function as if it were a method of the object.

But to understand this we first need to know the types of methods that exist.
### Types of methods
In Python [there are three types of methods](https://es.stackoverflow.com/questions/936/c%c3%b3mo-son-los-m%c3%a9todos-de-una-clase-python-por-defecto):

* *Instance methods*: receive the object's instance as their first parameter (they use `self`). By default, a class's methods in Python are instance methods.
* *Static methods*: receive nothing extra as their first parameter (no `self`); equivalent to free functions that could sit in the same translation unit in other languages.
* *Class methods*: receive the class as their first parameter; equivalent to methods marked static in other languages.




### 1. Instance methods:
+ Defined as methods of the class and require an instance of the class to be called.
+ Operate on the state of the specific instance.

Example:


````python
class Lib:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print(f"Hello, my name is {self.name}")

x = Lib("Carla")
y = Lib("Pedro")

# Method notation
x.greet()  # Prints "Hello, my name is Carla"
y.greet()  # Prints "Hello, my name is Pedro"
````

### 2) Static methods:
+ Defined as methods of the class, but do not require an instance of the class to be called; defined with the `@staticmethod` decorator.
+ Used as independent functions that operate on the class or its objects.

Examples:


````python
# https://stackoverflow.com/questions/30556857/creating-a-static-class-with-no-instances
# Static method
class Lib:
    def __init__(self):
        pass

    @staticmethod
    def method():
        print("Running method in Lib")

Lib.method() # Function notation

x = Lib()
x.method()   # Method notation
````

### 3) Class methods
Class methods are defined with the `@classmethod` decorator, and the word `cls` is used (which, like `self`, is not a reserved word but is widely used by the community). The `@classmethod` decorator is required to indicate that a method is a class method and that it should receive the `cls` parameter, for example
```python
class MyClass:
    ...
    @classmethod
    def method(cls, arg):
        return cls(arg)

MyClass.method("Hello")
```
in this example, `cls` is replaced by `MyClass` at run time. Note that `cls` is not the same as `self`, since:
* `self`: used for instance methods and refers to the current instance.
* `cls`: used for class methods and refers to the class itself.

Class methods are always decorated with `@classmethod`; this decorator is necessary to indicate that a method is a class method that should receive the `cls` parameter — without the decorator the method is interpreted as an instance method and an error results.

Silly example:


````python
# Class method
class Lib:
    @classmethod
    def hello(cls, name): # cls is used, but it could be another word
        print('Hello, %s!' % name)

Lib.hello('Carlos') # Hello, Carlos!
````

The following example creates a class that defines an object with two attributes: `x, y`, and then defines two methods: `__init__()` and `value()`. The third method, `Name`, is defined with the `@classmethod` decorator and creates a new instance of the class `Lib` with the specified value for attribute `x`, used as the initial value for the new instance; this method returns a new instance of the `Lib` class. Let's see:



````python
class Lib:
    def __init__(self, x, y=1000):
        self.x = x # Initial value of attribute x.
        self.y = y # Initial value of attribute y.

    def value(self):
        print('Hello, %s! the value is %d ' % (self.x, self.y))

    @classmethod
    def Name(cls, x): # Create an instance of the Lib class with value x.
        return cls(x)   # note the use of cls, though it could be another word.

# A new instance of the Lib class is created using the Name method.
c = Lib.Name('Carlos')

# The value method of instance c is called.
c.value()
````

Now we can see how to implement function and method notation using classes, examples:


````python
class Lib:
    def __init__(self,name):
        self.name = name

    @staticmethod
    def method1(): # Doesn't need self.
        print("Hello from the Lib class")

    def method2(self): # Does need self.
        print(f"Hello, my name is {self.name}")


# Function notation
Lib.method1()  # Prints "Hello from the Lib class".

# Method notation
x = Lib("Pedro")
x.method2()    # Prints "Hello, my name is Pedro".
````

````python
class Lib:
    def __init__(self, *args):
        self.args = args

    #@staticmethod       # Not allowed, errors on a.total().
    def total(self):
        return sum(self.args)

    @classmethod
    def obj(cls, *args):
        return cls(*args)

a = Lib.obj(1, 2, 3, 4) # Create object "a" with the tuple as an argument

print(Lib.total(a))      # Call the "total" method from the "Lib" class
print(a.total())         # Call the "total" method from the "a" object
````

````python
class Lib:
    def __init__(self, *args):
        self.args = args

    def __repr__(self):        # Returns the "args" tuple as a string: "(1, 2, 3, 4)"
        return repr(self.args) # https://codedamn.com/news/python/what-is-repr-in-python

    def total(self):
        return sum(self.args)

    def obj(*args):
        return Lib(*args) # Create an object of the Lib class with the "args" tuple as an argument


a = Lib.obj(1, 2, 3, 4) # Create object "a" with the tuple as an argument

print(Lib.total(a))      # Call the "total" method from the "Lib" class
print(a.total())         # Call the "total" method from the "a" object
print(a.__repr__())
````
