# Lists, tuples, dictionaries and arrays
A *list* is an ordered collection of elements (of the same or different type):
```python
    L = [1, "a", 3.14]  # list
```
A *tuple* is an ordered collection of elements (of the same or different type) whose elements cannot be changed once created. Tuples are written with their elements between parentheses, separated by commas,
```python
    T = (1, "a", 3.14) # tuple
```
You can convert a list to a tuple (or vice versa) with,
```python
    L = list(T)
    T = tuple(L)
```
In lists and tuples with `n` elements the index ranges from 0 to `n`-1, so for example,
```python
    L[0] = 1, L[1] = "a" and L[2] = 3.14.
```
*Dictionaries* are a data structure that stores an unordered set of pairs; to create one, the list of values is enclosed in braces, key-value pairs are separated by commas, and the key and value are separated by `:`, for example
```python
    D = {'x': 2, 'y': 1, 'z': 4}
```
they are used just like lists, for example,
```python
    D['x'] = 2
```
Note that in lists, tuples and dictionaries alike, the elements can themselves be other lists, tuples or dictionaries.

Finally, *arrays* are lists with special properties in which all elements are numbers; arrays belong to the `numpy` library, which is specialised for handling arrays efficiently — we will look at it later.

Example with lists:


````python
L = [0,10,20,40]

print(L[1])
print(L[1:3])     # print elements from index 1 to 3-1
print(L[::-1])    # gives the reversed list [40, 20, 10, 0]
print(L[0],L[-1]) # first and last elements of the list
[1,1]*3 # gives [1,1,1,1,1,1] and not [3,3] (if it were 3.0, a float, it would error)
````

## General properties
Below are some properties of lists,
```python
    n = len(L)   # number of elements
    item = L[i]  # take element i of the list or tuple
    seq = L[i:j] # take elements from i to j-1
```
Creating a list with elements,
```python
    L = []       # create an empty list, then elements can be added with append
    L = [expression for variable in sequence]
    L = list(sequence)
    L = list(expression for variable in sequence)
```
for example:
```python
    S = [x**2 for x in range(10)]    # means: S = {x² : x in {0 ... 9}}
    V = [2**i for i in range(13)]    # means: V = (1, 2, 4, 8, ..., 2¹²)
    M = [x for x in S if x % 2 == 0] # means: M = {x | x in S and x is even}
```
Modifying lists (not tuples, since they are immutable),
```python
    L.append(item)         # add item to the list; item is a single element
    L.extend(sequence)     # add sequence to the list; sequence has more than one element
    L.insert(index, item)  # insert item at a given index
```
Deleting elements from lists (not tuples),
```python
    del L[i]
    del L[i:j]             # delete elements from i to j-1
    item = L.pop()         # remove the last item of L and store it in item
    item = L.pop(0)        # remove the first item of L and store it in item
    item = L.pop(index)    # remove the item at index from L and store it in item
    L.remove(item)
```
Note `L.remove()` or `del L` deletes the whole list; in this case it does apply to tuples too.

The elements of a list can be another list,
```python
L2 = [['h','hello'], 1.0, 3.14]
```
to access the elements two indices are used,
```python
L2[0][0] = 'h'
L2[0][1] = 'hello'
L2[1]    = 1.0
L2[2]    = 3.14
```
Do not use `L2[i,j]`; that notation only works with numpy arrays.

## Functions
Functions in python are defined with the `def` keyword or the `lambda` keyword; the value returned by the function is given by the `return` statement — once python reaches this statement it exits the function and ignores the remaining lines of the function, which is why `return` is usually found at the end of a function (though it can appear earlier, guarded by an `if`). By definition a *routine* is a function without a `return` (for example, a function that only performs calculations and then uses `print` to display the results). This is best understood with an example:


````python
a = 1; b = 2

# 1) Define a function:
def f(x):
    y = a*x + b
    return y

# 2) or also
def f(x):
    if x < 2: return (a*x)
    else:     return (a*x + b)

# Use the function:
y = f(3)       # the value of f(x) can be assigned to a variable

y
````

A function can return several values at once (arrays, tuples, dictionaries, etc.),


````python
def f2(x):
    y = a*x + b
    z = 3*x
    return y, z

Y, Z = f2(2) # must be assigned to a tuple
Y, Z         # print the values of the tuple
````

### Passing variables to functions
In general, when a variable is passed to a function, python creates a copy of it (this is called "*pass by value*"), so it doesn't matter what letter or name is used to define it, for example,


````python
x = 10    # the variable x should not change when the function is called.
def f(x): # this x has nothing to do with the previous x.
    return 4*x

z = 2     # note that z, not x, is used
y = f(z)
x, y      # note f(z) does not change the value of x
````

### Functions of several variables with predefined parameters
You can also define a function of several variables and assign default values (or initialise) the variables; in the following example the function $g(x,a,b)=ax+b$ is defined, but the parameters $a,b$ take the default values (2,1),


````python
# function of three variables
def g(x, a=2, b=1):
    return (a*x + b)

#  using variables in functions and their initialisation
y  = g(3)      # x is initialised and a,b take their default values
y1 = g(3,4,7)  # all 3 values are initialised
y2 = g(3, a=5) # the initial value of a is changed
y3 = g(3, b=2) # the initial value of b is changed

y, y1, y2, y3  # show the values as a tuple
````

In the following example the `lambda` keyword is used to define $f(x)$ and $g(x,y)$:


````python
# Lambda functions of one and two variables
f = lambda x: x**2 + 3
g = lambda x,y: x*y
y = f(3)
g(2,y)  # function of two variables
````

Another *example*: if $f(x)=x^2$ and $g(x)=2x+1$, compute $f(g(x))$ for $x=1$.


````python
g = lambda x: x*2 + 1
f = lambda x: x**2

f(g(1))
````

### Functions returning functions
A function can also return another function; let's see,


````python
# How to return a function from another function

# 1) one way
def cylinder(r):
    def volume(h):
        return 3.1416*r*r*h
    return volume                # returns a function

# 2) or also
def cylinder(r):
    return lambda h: 3.1416*r*r*h # returns a function


new_function = cylinder(10) # create a new function
new_function(5)              # use the new function => 1570.8
````

### Variable values and their scope

Variables are global by default (i.e. every function or routine in the program can see and use the values stored in them), but if a function is given a variable with the same name and a different value, python creates a local copy with the new value (unless the function explicitly declares the variable as global), and this copy is only defined inside the function. Example:


````python
h = 20
print("hello before",h)
def f():
    # Here h is a local variable, but
    # global h # uncommenting this line makes h global.
    h = 5
    print ("hello inside the function", h) # local variable

f() # print
print("hello after",h)
````

# How to import libraries
there are three ways,

1) Import only a single method of interest from the library (where a method is a function defined in the library),
```python
from library import method
```
then the function is used as follows:
```python
var = method()
```
Advantage: there is no need to reference the library the method comes from.
Disadvantage: without referencing the library, you lose the context of where the method comes from.

2) You can also import the whole library:
```python
import library
```
but in this form you have to say where the method comes from; more precisely you have to give the full path to the method when creating the object with variable var:
```python
var = library.method()
```
3) but you can create an alias:
```python
import library as lib
```
this simplifies creating the object:
```python
var = lib.method()
```
4) Finally you can also write:
```python
from library import *
```
this way a reference to every object in the library is created, as in 1) but importing everything; the problem is that the library's origin is lost when the method is created,
```python
var = method()
```
this is not convenient since two libraries may have functions with the same name; this is the case, for example, with the trigonometric functions in numpy and sympy — there is no way to know if `sin(x)` is `numpy.sin(x)` or `sympy.sin()`, and worse still, if we import as follows,
```python
from numpy import *
from sympy import *
```
since only the methods are imported without the library names, the second import overwrites the methods of the first: the result is that only the symbolic trigonometric functions remain, so this style is only recommended for short scripts whose meaning is explicit, and it is also useful for interactive use in IPython.

### Passing arrays to functions
Arrays and dictionaries — unlike plain variables — if passed as function arguments can be permanently modified inside the function (this is called "*pass by reference*" and does not create a copy). This can be used to return arrays, but if the modification is done via numpy functions, these create copies of the array inside the function and the original arrays are not permanently modified; let's see,


````python
import numpy as np
a = np.zeros(10)
b = np.zeros(10)

def f(a, b):
    for i in range(10): a[i] = 3 # permanently modifies the value of array a.
    b = np.ones(10)*5            # does not permanently modify array b's value,
                                 # creates a copy of b inside the function (a bug??).
    print(a,b)                   # print the modified arrays.

f(a,b)  # call the function.
a, b    # Note that only a is permanently modified; only a can be used as the return of f(a,b).
````

````python
# ----------------- functions with arrays ----------------------------
root = np.sqrt(x1)    # array whose elements are the square roots of the elements of x1
print (root)          # equivalent to x1**.5.

y = np.sin(x1)        # array where each element is the sine of each element of x1
print (y)

dd = np.dot(x1,x5)    # dot product (scalar product) of x1 and x2
print (dd)

magnitude = x1.dot(x1) # same result as dot(x1,x1); this is the magnitude of vector x1
print (magnitude)
````

## Saving data to a file

There are several ways to do this; the most standard is with `open`,
```python
file = open(filename,"w")    # open a file for writing
file.write("value: %s "%a)   # write variable a to the file
    ...                      # do something
file.close()                 # close the file at the end of the program
```
the first argument is the file to read or write, and the second argument depends on whether it's for writing or reading,

* <font color='brown'>'r'	</font> means the file is opened for reading.
* <font color='brown'>'w'	</font> is a file opened for writing (if the file exists it will be overwritten).
* <font color='brown'>'a'	</font> opens a file and, if it has existing data, new data is appended at the end of the file.
* <font color='brown'>'r+'</font>	open a file for reading and writing.
* <font color='brown'>'rb'</font>	open a file in "comma separated file" (csv) format (requires `import csv`) (b is for binary files, e.g. images).
* <font color='brown'>'wb'</font>	open a file for writing in csv format.

If it's a numpy array there's an easier way,
```python
    np.savetxt('Table.dat',myarray, fmt=format, header='')
```
with this function there is no need to close the file, where `fmt` is the format to use, for example `fmt = '%14.10f'`, meaning 14 spaces total with 10 after the decimal point; the `f` means the data are real, floating-point numbers.


````python
# example with open
x = y = z = np.arange(0.0,5.0,1.0) # data

file = open("test.dat","w") # open a file for writing

for i in range(len(x)):
    file.write("%7.3f %7.3f %7.3f\n"%(x[i], y[i], z[i]))

file.close()                # close the file at the end of the program

%cat test.dat               # view the file created in your directory (Linux only)
````

````python
# example, print without a format
x = y = z = np.arange(0.0,5.0,1.0)

np.savetxt('test.dat', list(zip(x,y,z)))         # way 1
%cat test.dat # view the data from the created file
````

````python
# tidier, printing with a format
np.savetxt('test.dat', np.c_[x,y,z], fmt='%3d %2d %2d', header='x, y, z') # way 2
%cat test.dat # view the data from the created file (Linux only)
````

## Loading data from a file
There are also several ways to do this; the most usual is to do the same as for saving files (as above, but changing `"w"` to `"r"`), but if the `with` command is used there is no need to close the file at the end of reading (note that `with` also works for writing),


````python
with open("test.dat") as file:  # "file", variable that stores the data
    #data = file.read()         # read the whole file,
    data = file.readlines()     # or read line by line.
    #do something with the data.

for line in data:
    words = line.split()        # split the line string into substrings
    print(words)                # note that it prints strings, not numbers
````

But even easier: if the file only contains numbers or a data table, the `numpy.genfromtxt()` command can be used to import the file,
```python
     E = np.genfromtxt("test.dat")    #  import a 2D array
```
where `E[0], E[1], ..., E[n]` are the rows of the file; if you want the columns, transpose the matrix `E`. Lines starting with `#` are ignored. Note that `np.loadtxt()` also exists and is similar (see the manual); examples:


````python
# Example 1
E = np.genfromtxt("test.dat")
E = E.T # Transpose to obtain: E[0] column 1, E[1] column 2, E[2] column 3
print (E[0], E[1])
E
````

````python
# example 2, the more practical way
x,y,z = np.genfromtxt("test.dat",usecols=(0,1,2),unpack=True)
print(x,y,z)

%rm test.dat # delete the file (Linux, Unix only)
````
