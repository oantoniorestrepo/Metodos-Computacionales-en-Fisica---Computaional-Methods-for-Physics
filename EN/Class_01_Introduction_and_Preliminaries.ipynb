# Introduction to programming


### What is a problem in computer science
A problem is a yes/no question about a particular input, for example:

    given inputs a,b,c, is a + b = c?

that is, a decision must be made; this is known as a *decision problem*.

### What is an algorithm
An algorithm is a set of logical steps that answers the question in a bounded amount of time.

### What is a program
A program is a set of logical instructions that solve a given problem and are stored in a file.

### Decidable and undecidable problems
A problem is decidable if it has an algorithmic solution in finite time; if no algorithm can answer the problem in finite time, the problem is undecidable.
Are all problems decidable? In 1931 the mathematician [Kurt Gödel](https://en.wikipedia.org/wiki/Kurt_G%C3%B6del) proved that they are not (see [Gödel's incompleteness theorem](https://www.gaussianos.com/que-dice-exactamente-el-primer-teorema-de-incompletitud-de-godel/)). Later [Alan Turing](https://en.wikipedia.org/wiki/Alan_Turing), in his [thesis](https://math.stanford.edu/~feferman/papers/turing.pdf), proved the analogous result in computer science by formulating the [halting problem](https://en.wikipedia.org/wiki/Halting_problem), which refers to the impossibility of constructing a program (or algorithm) that determines whether an arbitrary program, run on a Turing machine, halts or runs forever (infinite loop); this problem is undecidable. If a problem is undecidable, there is no point spending effort looking for a solution — it is simply set aside.

### Problems of type P and NP
Decidable problems are split into type P (polynomial) and type NP (non-polynomial). This criterion is based on how the algorithm's running time scales with the number of operations, so if the running time of an algorithm to obtain.

## Programming languages
Programming languages are classified as low, medium or high level relative to *assembly*, which is the language computers natively understand. A high-level language is one that is closer to natural human language; it is more practical and easier for the programmer to work with and carries fewer risks of mistakes, although the resulting program runs more slowly.

Low/medium-level languages: these are compiled, i.e. a compiler is used to translate human-readable instructions into machine language, which is binary (zeros and ones). The most common are:

Python can be used in two ways: "interactive" mode or "script" mode; let's see both:


### Python in interactive mode
In interactive mode each Python instruction (or command) can be executed and its result seen immediately. For this you open `IDLE`, Python's default terminal; on Linux you simply type `python` in a terminal (opens a bare terminal) or, better, IPython (a smarter terminal); on Windows explicitly look for IDLE. For example, open IDLE and type:
```python
In [1]: print('hello world') # this is a comment
```
You can also run simple command lines in a terminal without opening IDLE, for example:
```bash
$]  python -c 'print("hello world")'
```
Note that the instruction goes in single quotes and "hello world" goes in double quotes to avoid ambiguity.

### Python "scripts"
You can also work in the traditional way: write a "script" or program in any editor and then call the program from a terminal.
A Python script is a program saved to a file with a `.py` extension, e.g. "my_program.py".
It's important to save programs with names that reflect what the script does, so it's easier to find and remember what it is for.

To run the script from a Linux shell terminal (`cmd` on Windows) we write:
```bash
$]  python my_program.py
```
and in an ipython terminal we use the magic command `%run`:
```python
In [1]: %run my_program.py
```

## Logic and conditionals: and, or, not
Conditionals are an essential part of programming since they allow decisions to be made in order to answer the problem; in python they are defined as,
```python
if (condition): do something
else: do something else
```
another way
```python
if (condition):
     do something
else:
     do something else
```
The `condition` must be a logical value taking either False or True.
Given the above, note for example that,
```python
if (x<10): y=2; print('hello'); print(x)
```
is equivalent to writing,
```python
if (x<10):
    y=2
    print('hello')
    print(x)
```

If there are several conditions they are implemented via the `elif` instruction:
```python
if  (condition1):
elif(condition2):
else:
```
If there are more conditions, more `elif` clauses are added.
By default the operation is carried out if the condition is true; otherwise it is skipped and execution moves to the next line. It is important to note that computers only understand zeros and ones, so by convention `False` has the value zero and `True` the value one (or any other non-zero number); with this in mind, what does the following print?

### Truth tables in computing
The truth tables given in mathematical logic,

p | q |(not p) |(not q) |(p and q) | (p or q)
--|---|---|---|---|---
T | T | F | F | T | T
T | F | F | T | F | T
F | T | T | F | F | T
F | F | T | T | F | F

and defined by their logical operations `and`, `or`, `not`, are very important for understanding programming and decision-making. For instance, if we define `p = 1<3` and `q = 3>4`, which take the values `p = True` and `q = False`, we can build different logical conditions using the truth tables. Example:


````python
p = 1<3; q = 3>4 # gives p = True and q = False
print ('values of p and q: ', p, q)
np = not p
print ('negation of p:', np )

# using truth values in conditionals
if (p and q): print ('"p and q" is true')
if (p or q): print ('"p or q" is true')
if (not(p and q)): print ('negation of "p and q" is true')
````

Several conditions can be combined, for example,
```python
if (a > x and x < b or y == d):
```
Truth tables can be used to rewrite a conditional; recall from logic that negating a conditional swaps `and` for `or` and reverses the $>$ and $<$ operators, i.e.
```python
not (a > b) is equivalent to (a < b)
not (a > x and x < b) is equivalent to (a < x or x > b)
```
and so on. Given the above, what is the negation of?

### Nested ifs
Suppose you have several conditions; you could write something like
 ```python
 if (condition1 and condition2 and ... and conditionN):
 ```
 which only runs if all the conditions are true. A more practical way of implementing the same thing is to use several nested `if` statements, i.e.
```python
if condition1:
    if condition2:
        ...
            if conditionN: print(x)
```
for computational efficiency the most restrictive condition should go in the first `if`, so that if it fails, the remaining `if` statements and their operations are never even checked.





# For and while loops
This is the second essential part of programming after conditionals. When we want an operation to repeat many times, we use loops. The syntax of the `for` loop is,
```python
    for var in array: do something
```
in the `for` loop `range(N)` is usually used (in python 2.7 `xrange(N)` was used; this command no longer exists in python 3.x) to build a list of numbers that the `for` loop iterates over. For `while` the syntax is,
```python
    while (condition): do something
```
to interrupt the loop the `break` command is used, or to skip a step the `continue` command is used, example:


````python
for i in range(6): # range(start=0, stop=6-step, step=1)
    print('variable', i, i*2, i**2)
````

````python
i = 0                   # initialisation
while i < 6:            # run the loop
    if i == 3: break    # stopping condition
    i += 1              # increment the value by one
    #i = i + 1          # equivalent to previous line
    if i == 2: continue # skip 2 (use "==" to compare values; "is" compares
                        # object identity, not equality, and should not be used here)
    print(i)
````

### Using enumerate
Consider the following `for` loop, where the counter over the number of iterations is needed, but `range` increases in steps of two (i.e. `range(2,10,2)`). This is especially important if the counter is, say, the index of a list or array (see the definition of list below); to solve this we would need to create an extra variable besides `i` for the counter,
```python
    i=0
    for j in range (2,10,2):
        x = i*j + 1
        print (i, j, x)
        i = i + 1 # increment i by 1
```
The `enumerate(iterable, start=0)` method is another option for the `for` loop and adds a counter to the iteration more easily; the following example does the same as the previous loop:


````python
# Example 2
for i,j in enumerate(range(2,10,2)): print(i,j)
````

````python
# example 3
for i,j in enumerate(range(10),5): print(i,j)
````
